# TomoAcquire User Tutorial

The TomoAcquire library is a Tomobase plugin which provides users with the ability to remotely control a periphery devices e.g. microscopes, holders, detectors. etc. To use Tomoacquire the connected device must be able to respond to messages provided by the TomoAcquire plugin. 

The purpose of this guide is to show users how to use the Tomoacquire plugin to perform a tomographic acquisition. As this is an extension to Tomobase, it is recommended to see the Tomobase totorial first. 

## 1. Device Connection & Setup

```python
import logging
import tomobase
tomobase.bootstrap(jupyter_enabled=True)

from tomobase import phantoms, procedures, data_classes, logger
from tomobase import jupyter

logger.setLevel(logging.VERBOSE)
jupyter.LogWidget().open() 
```

```python
from tomoacquire.core import base_classes, protocols
from tomoacquire import devices

microscope = base_classes.DeviceController(protocol=protocols.TomoAcquireProtocol, address="tcp://localhost", req_port=50000, reply_port=50001, key="keyfile.dat")
microscope.components.register("Column", devices.ColoumnController)
microscope.components.register("Stage", devices.StageController)
microscope.components.register("Detectors", devices.STEMDetectorController)

#microscope.components.register("DENSHeater", devices.DENSHeaterController(protocols.TomoAcquireProtocol, address="tcp://localhost", req_port=50000, reply_port=50001, key="keyfile.dat"))
microscope.save()
```
